```python
def build_profile(**kwargs):
    profile = {}
    for key, value in kwargs.items():
        profile[key] = value
    return profile
user = build_profile(name = "solomon", Age = 18, height = 4.5, city = "Obikabia", role = "d.J",)
print(user)
```

```python
def develop_details(**Favour):
    """build a  user profile from any number of named fields"""
    profile = {}
    for key, value in Favour.items():
        profile[key] = value
    return profile

user = develop_details(name = "Gift",  Age = 23,  city = "Aba")
print(user)
```
